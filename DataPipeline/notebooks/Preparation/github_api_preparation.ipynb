```python
# Setup GraphQL Client
from aiographql.client import GraphQLClient
import os
client = GraphQLClient(endpoint="https://api.github.com/graphql", headers={"Authorization": f"Bearer {os.getenv('GITHUB_TOKEN')}"})
```

```python
# Build the schema if have 
from graphql import build_ast_schema, parse


schema_path = "/Users/nguyen.truong.an/Personals/copilot-metrics-viewer/DataPipeline/src/common/schema.docs.graphql"
if schema_path:
    with open(schema_path, "r", encoding="utf-8") as schema_file:
        schema_str = schema_file.read()
    schema_ast = parse(schema_str)
    client.schema = build_ast_schema(schema_ast)
```

```python
# Draw all repositories
query = """
        query($owner: String!, $name: String!) {
            repository(owner: $owner, name: $name) {
                id
                name
                nameWithOwner
                description
                url
                homepageUrl
                createdAt
                updatedAt
                pushedAt
                isFork
                isArchived
                isDisabled
                isLocked
                isPrivate
                hasIssuesEnabled
                hasWikiEnabled
                hasDiscussionsEnabled
                primaryLanguage {
                    name
                    color
                }
                languages(first: 10) {
                    nodes {
                        name
                        color
                    }
                }
                defaultBranchRef {
                    name
                    target {
                        ... on Commit {
                            oid
                            committedDate
                        }
                    }
                }
                stargazerCount
                forkCount
                watchers {
                    totalCount
                }
                issues {
                    totalCount
                }
                pullRequests {
                    totalCount
                }
                diskUsage
                licenseInfo {
                    name
                    spdxId
                }
            }
        }
        """

response = await client.query(query, variables={"owner": "moneyforward", "name": "received_invoice"})
print(response.data)
```

```python
import asyncio
import nest_asyncio
nest_asyncio.apply()
```

```python
# First, let's create a GraphQL query for organization repositories
query = """
query($org_name: String!, $limit: Int!) {
    organization(login: $org_name) {
        repositories(first: $limit, orderBy: {field: UPDATED_AT, direction: DESC}) {
            nodes {
                id
                name
                url
                createdAt
                updatedAt
                pushedAt
                isPrivate
                isArchived
            }
            pageInfo {
                hasNextPage
                endCursor
            }
            totalCount
        }
    }
}
"""

# Execute the query
async def get_org_repositories():
    response = await client.query(
        query, 
        variables={
            "org_name": "moneyforward",  # Replace with your organization name
            "limit": 100  # Adjust the limit as needed
        }
    )
    return response.data

# Run the query

result = asyncio.run(get_org_repositories())
print(result)
```
