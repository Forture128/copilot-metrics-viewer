## Import the required libraries

````python
# Import Required Libraries
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import json
import os
import numpy as np
````

### Calculate Cumulative Team Metrics

*Based on the data Json which {team_tag}_metrics*

````python
# Load and prepare data
# Copilot Usage:  f"{OUTPUT_DIR}/github_team_{team_tag}_metrics.json",
# Team members: f"{OUTPUT_DIR}/github_team_{team_tag}_members.json
# Team repos:   f"{OUTPUT_DIR}/github_team_{team_tag}_repos.json


# Collect all metrics data of all teams
data_path = os.path.join("..", "..", "data", "raw")
# "*_metrics.json"
metrics_files = [f for f in os.listdir(data_path) if f.endswith("_metrics.json")]
members_files = [f for f in os.listdir(data_path) if f.endswith("_members.json")]

# Load all metrics data
metrics_data = {}
for file in metrics_files:
    with open(os.path.join(data_path, file), "r") as f:
        # file: github_team_adiv_metrics.json => github_team_{team_tag}_metrics.json
        # Set team_tag as a key in the dictionary
        team_tag = file.split("_")[2]
        metrics_data[team_tag] = json.load(f)
# Transform the metrics_data dictionary into a DataFrame
metrics_dfs = []
for team_tag, data in metrics_data.items():
    # Flatten the nested JSON structure
    df = pd.json_normalize(data)
    # Add a column for team_tag
    df['team_tag'] = team_tag
    metrics_dfs.append(df)

# Concatenate all the DataFrames into a single DataFrame
metrics_df = pd.concat(metrics_dfs, ignore_index=True)

######################
# Load all members data
members_data = {}
for file in members_files:
    with open(os.path.join(data_path, file), "r") as f:
        team_tag = file.split("_")[2]
        members_data[team_tag] = json.load(f)

# Transform the members_data dictionary into a DataFrame
members_dfs = []
for team_tag, data in members_data.items():
    # Flatten the nested JSON structure
    df = pd.json_normalize(data)
    # Add a column for team_tag
    df['team_tag'] = team_tag
    members_dfs.append(df)
members_dfs = pd.concat(members_dfs, ignore_index=True)  

# # Load all repos data
# with open(os.path.join(data_path, "repositories_page_1.json"), "r") as f:
#     repos_data = json.load(f)

````

````python
import plotly.graph_objects as go

# Calculate the acceptance rate
metrics_df["acceptance_rate"] = (metrics_df["total_acceptances_count"] / metrics_df["total_suggestions_count"]) * 100

# Fix team_tag column if it too long
metrics_df["team_tag"] = metrics_df["team_tag"].apply(lambda x: x[:15])

# Aggregate the data by team
team_metrics = metrics_df.groupby('team_tag').agg({
    'total_suggestions_count': 'sum',
    'total_acceptances_count': 'sum',
    'acceptance_rate': 'mean'
}).reset_index()

# Sort the DataFrame by acceptance rate
team_metrics_sorted = team_metrics.sort_values(by="acceptance_rate", ascending=False)

# Create a bar chart for total suggestions and acceptances by team_tag
fig = go.Figure()

# Add bar trace for total suggestions
fig.add_trace(go.Bar(
    x=team_metrics_sorted["team_tag"],
    y=team_metrics_sorted["total_suggestions_count"],
    name='Total Suggestions',
    marker_color='rgba(75, 192, 192, 0.6)',
))

# Add bar trace for total acceptances
fig.add_trace(go.Bar(
    x=team_metrics_sorted["team_tag"],
    y=team_metrics_sorted["total_acceptances_count"],
    name='Total Acceptances',
    marker_color='rgba(255, 99, 132, 0.6)',
))

# Add a line chart for the acceptance rate
fig.add_trace(go.Scatter(
    x=team_metrics_sorted["team_tag"],
    y=team_metrics_sorted["acceptance_rate"],
    name='Acceptance Rate',
    marker_color='rgba(54, 162, 235, 0.6)',
    yaxis='y2'
))

# Update layout for dual y-axis
fig.update_layout(
    title='Total Suggestions, Acceptances, and Acceptance Rate by Team',
    xaxis=dict(title="Teams"),
    xaxis_tickfont_size=14,
    yaxis=dict(
        title='Count',
        tickfont_size=14,
    ),
    yaxis2=dict(
        title='Acceptance Rate (%)',
        titlefont_size=16,
        tickfont_size=14,
        overlaying='y',
        side='right'
    ),
    legend=dict(x=0.1, y=1.1, orientation='h'),
    barmode='group',
    # bargap=0.15, # gap between bars of adjacent location coordinates.
    # bargroupgap=0.1 # gap between bars of the same location coordinates.
)

fig.show()
````

### Visualize Total Suggestions and Acceptances
Use Plotly to create a bar and line chart for total suggestions and acceptances over time by select team.

**Description**: Calculate the acceptance rate as a percentage of total suggestions accepted over time. This metric can show how useful and relevant Copilot suggestions are over time.

- **Visualization**: Use a line chart to show the acceptance rate over time. You can also create a comparison between the acceptance rates of suggestions and chat interactions and the columns chart for total suggestions and acceptances.
- **Calculation**: 
    ```
        daily_metrics["acceptance_rate"] = (daily_metrics["total_acceptances_count"] / daily_metrics["total_suggestions_count"]) * 100
    ```

````python
# Create a dropdown to select different teams
team_tags = metrics_df['team_tag'].unique()

# Create a figure
fig = go.Figure()

# Add traces for each team
for team_tag in team_tags:
    team_data = metrics_df[metrics_df['team_tag'] == team_tag]
    
    fig.add_trace(go.Bar(
        x=team_data["day"],
        y=team_data["total_suggestions_count"],
        name=f'Total Suggestions ({team_tag})',
        visible=False
    ))

    fig.add_trace(go.Bar(
        x=team_data["day"],
        y=team_data["total_acceptances_count"],
        name=f'Total Acceptances ({team_tag})',
        visible=False
    ))

    fig.add_trace(go.Scatter(
        x=team_data["day"],
        y=team_data["acceptance_rate"],
        name=f'Acceptance Rate ({team_tag})',
        line=dict(width=2),
        yaxis='y2',
        visible=False
    ))

# Make the first team's data visible
for i in range(3):
    fig.data[i].visible = True

# Create dropdown buttons
dropdown_buttons = []
for i, team_tag in enumerate(team_tags):
    dropdown_buttons.append(
        dict(
            label=team_tag,
            method="update",
            args=[{"visible": [False] * len(fig.data)},
                  {"title": f"Total Suggestions, Acceptances, and Acceptance Rate for {team_tag}"}]
        )
    )
    for j in range(3):
        dropdown_buttons[-1]["args"][0]["visible"][i * 3 + j] = True

# Add dropdown to the figure
fig.update_layout(
    updatemenus=[
        dict(
            active=0,
            buttons=dropdown_buttons,
            x=0.17,
            xanchor="left",
            y=1.15,
            yanchor="top"
        )
    ]
)

# Update layout for dual y-axis
fig.update_layout(
    title='Total Suggestions, Acceptances, and Acceptance Rate by Team',
    xaxis_tickfont_size=14,
    yaxis=dict(
        title='Count',
        titlefont_size=16,
        tickfont_size=14,
    ),
    yaxis2=dict(
        title='Acceptance Rate (%)',
        titlefont_size=16,
        tickfont_size=14,
        overlaying='y',
        side='right'
    ),
    legend=dict(x=0.1, y=1.1, orientation='h'),
    barmode='group',
)

fig.show()
````
